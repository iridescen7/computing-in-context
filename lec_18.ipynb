```python
import pandas as pd

url = "https://storage.googleapis.com/python-public-policy2/data/311_requests_2018-19_sample.csv.zip"

df = pd.read_csv(url, compression="zip")

df.head()
```

```python
agency = df["Agency"]

agency
```

```python
counts = df["Descriptor"].value_counts()
counts
```

```python
groups = df.groupby(["Agency", "Descriptor"]).size()
counts_per_agency = groups.reset_index(name="Count").sort_values(
    ["Agency", "Count"], ascending=False
)
counts_per_agency
```

```python
counts_per_agency.drop_duplicates("Agency").sort_values("Agency")
```
