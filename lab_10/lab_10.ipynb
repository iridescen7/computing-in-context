### Step 3

Use [`merge()`](https://pandas.pydata.org/docs/user_guide/merging.html#merge) to combine the two, and output the resulting table.


Prepare for merging.

```python
pop["Borough"].unique
```

```python
aed["Borough"].value_counts(dropna=False)
```

According to the Column Information - X: Bronx, B: Brooklyn, M: Manhattan, Q: Queens, R: Staten Island

Renaming.

```python
def recode_borough(row):
    borough = row["Borough"]

    if pd.isna(borough):
        return "Unknown"
    elif borough == "M":
        return "Manhattan"
    elif borough == "X":
        return "Bronx"
    elif borough == "B":
        return "Brooklyn"
    elif borough == "Q":
        return "Queens"
    elif borough == "R":
        return "Staten Island"


aed["Borough"] = aed.apply(recode_borough, axis=1)
```

```python
aed["Borough"].value_counts(dropna=False)
```

Modifying variable formatting.

```python
pop["2020"] = pop["2020"].astype(str).str.replace(",", "").str.strip().astype(int)

pop["2020"].dtype
```

```python
pop["Borough"] = pop["Borough"].str.strip()
```

### Step 4

Using the two datasets above, use pandas to produce an aggregate per-capita statistic by borough.

The dataset you chose before may not work for this. That's fine, pick another.

#### Hint

You're creating a "number of [thing] per capita by borough" table.

1. Do a [`groupby()`](https://pandas.pydata.org/docs/user_guide/groupby.html) on the original dataset.
1. Join with the populations by borough.
1. Compute the per-capita values as a new column.


Because the data is about AED provisions, here I calculated the number of per capita AED allocations for each borough.

First calculate the counts of AED for each borough.

```python
aed_counts = aed.groupby("Borough").size().reset_index(name="AED_records")
aed_counts
```

Merging by Borough name.

```python
result = aed_counts.merge(pop[["Borough", "2020"]], on="Borough", how="inner")
result
```

Calculate per capita data.

```python
result["records_per_100k"] = (result["AED_records"] / result["2020"] * 100000).round(2)
result
```

Therefore, we can see that per capital AED allocation is highest for Manhattan, then Brooklyn and Queens, while Staten Island is the lowest.
